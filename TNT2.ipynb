```python
from anywidget import AnyWidget
import glasbey as gl
import ipywidgets as wg
from jscatter import Scatter
from matplotlib.colors import ListedColormap
import numpy as np
import os
import pandas as pd
from scipy.stats import multivariate_normal
from sklearn.covariance import EmpiricalCovariance
from sklearn.datasets import make_blobs
import traitlets as tl

import thisnotthat as tnt
import thisnotthat._thisnotthat as _tnt
```

```python
blobs, labels = make_blobs(20000, centers=10)
noise = multivariate_normal(mean=np.mean(blobs, axis=0), cov=EmpiricalCovariance().fit(blobs).covariance_).rvs(1000)
df = pd.DataFrame(
    data=np.vstack([blobs, noise]),
    columns=["x", "y"]
).assign(
    label=pd.Categorical(np.concat([labels, np.full((noise.shape[0],), -1)]), categories=np.unique(labels)),
    u=np.random.random((labels.shape[0] + noise.shape[0],))
).astype({"label": "category"})
df
```
